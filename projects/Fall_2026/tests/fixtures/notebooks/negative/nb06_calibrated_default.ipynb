```python
# Framework — do not edit
from clinical_ml_kits import load_kit
kit = load_kit("toy_binary", data_source="synthetic")
data = kit.load()
```

```python
# Baseline — working example
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from clinical_ml_kits.preprocessing import make_preprocessor
baseline = Pipeline([("preprocess", make_preprocessor(kit.feature_spec)), ("model", LogisticRegression(C=1.0, max_iter=1000))])
baseline.fit(data.train.X, data.train.y)
val_proba = baseline.predict_proba(data.validation.X)[:, 1]
```

```python
# Your work
from sklearn.calibration import CalibratedClassifierCV
cal = CalibratedClassifierCV(baseline, method="sigmoid")
cal.fit(data.train.X, data.train.y)
```
